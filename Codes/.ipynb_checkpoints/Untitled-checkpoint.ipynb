**Paramètres physiques**

```python
######################
hbar=1.0546e-34
c=2.997e8
q=1.60217653e-19 
temp=300
kappa=1.3806622e-23
Nq=3
```

```python
#####################
UT=kappa*temp/q
```

**Levels quantum**

```python
#####################
Omega_gap=1.18e15
Omegap=Omega_gap*np.array([-1.0,0.0,3.0])

Ep=hbar*np.diag(Omegap)
Energ=-np.diag([Omegap[0]-Omegap[1],Omegap[0]-Omegap[2],Omegap[1]-Omegap[2]])/hbar
```

**Pauli relaxation**

```python
##############
Gam=-1.e10
Dg=-np.diag([Gam]*3)
```

**Electromagnetic field**

```python
E0=4.2186e9
####
#####################
M=np.array([9.1e-29,9.1e-29,9.1e-29])
Minf=max(abs(M))

#####################
Omeg_Rabi=E0*Minf/hbar
```

**Coulomb parameters**

```python
R=1.e-24
R0=-np.diag([0,2*R,2*R])/hbar
R0,Energ
```

**Numerical paramaeters**

```python
Nt=100000
dt=1/Nt

#####################
Omeg_Rabi=E0*Minf/hbar
```

**Initial density**

```python
Uinit=np.array([[0,0,1]+[1]*6]).T
Uinit
```

**Matrix**

```python
def AL(sol):
    resu=np.zeros((9,9))
    ####Add free energies
    resu[3:6,6:]=-Energ/(E0*Minf)
    resu[6:,3:6]=Energ/(E0*Minf)
    ####Add relaxation in coherences
    resu[3:6,3:6]=Dg/Omeg_Rabi
    resu[6:,6:]=Dg/Omeg_Rabi

    return resu@sol
```

**Functions**

```python
def sol_RK4(dtime,Ntime):
    sol=Uinit
    for t in range(1,Ntime):
        ####
        R1=sol+dtime*AL(sol)/2
        R2=sol+dtime*AL(R1)/2
        R3=sol+dtime*AL(R2)
        #####
        sol=sol+dtime*(AL(sol)+2*AL(R1)+2*AL(R2)+AL(R3))/6
    

    print(sol)
    
```

```python
sol_RK4(dt,Nt)
```
